3. Handling Missing Data

```python
# Checking Missing Values

df.isnull().sum()
```

```python
# Calculating Percentage of Missing Values

missing_percentage = (df.isnull().sum() / len(df)) * 100
missing_percentage
```

```python
#The TotalCharges column contains blank string values instead of actual null values.These blank values must be converted into missing values before further processing.

# Checking Blank Values in TotalCharges Column

(df["TotalCharges"] == " ").sum()
```

```python
# Replacing Blank Values with NaN

df["TotalCharges"] = df["TotalCharges"].replace(" ", np.nan)
```

```python
# Checking Missing Values After Replacement

df.isnull().sum()
# After replacing blank strings with NaN values, the missing values in the TotalCharges column are correctly identified.
```

```python
# Checking Data Type of TotalCharges

df["TotalCharges"].dtype
```

```python
# Converting TotalCharges to Numeric

df["TotalCharges"] = pd.to_numeric(df["TotalCharges"])
```

```python
# Removing Rows Containing Missing Values

df.dropna(inplace=True)
```

```python
# Verifying Missing Values After Cleaning

df.isnull().sum()
```

5. Data transformation

```python
#The Churn column contains Yes and No so will convert it into numerical values.
```

```python
# Encoding Churn Column

df["Churn"] = df["Churn"].map({"Yes": 1, "No": 0})

df["Churn"].head()
```

```python
# Encoding Gender Column

df["gender"] = df["gender"].map({"Male": 1, "Female": 0})

df["gender"].head()
```

```python
# Encoding Binary Features

binary_columns = [
    "Partner",
    "Dependents",
    "PhoneService",
    "PaperlessBilling"
]

for column in binary_columns:
    df[column] = df[column].map({"Yes": 1, "No": 0})

df[binary_columns].head()
```

```python
# Checking Data Types After Transformation

df.dtypes
```

6. Univariate analysis with a minimum of three visualizations

Visualization 1: Distribution of Monthly Charges

```python
# Plotting Distribution of Monthly Charges

plt.figure(figsize=(10,5))
sns.histplot(df["MonthlyCharges"], bins=32, kde=True, color="skyblue")
plt.title("Distribution of Monthly Charges")
plt.xlabel("Monthly Charges")
plt.ylabel("Frequency")
plt.show()
```

Visualization 2: Customer Churn Distribution

```python
# Plotting Customer Churn Distribution

plt.figure(figsize=(10,5))
sns.countplot(x="Churn", data=df, palette="Set2")
plt.title("Customer Churn Distribution")
plt.xlabel("Churn")
plt.ylabel("Number of Customers")
plt.xticks([0,1],["No","Yes"])
plt.show()
```

Visualization 3: Distribution of Customer Tenure

```python
# Plotting Distribution of Customer Tenure

plt.figure(figsize=(20,5))
sns.boxplot(x=df["tenure"], color="lightgreen")
plt.title("Distribution of Customer Tenure")
plt.xlabel("Tenure (Months)")
plt.show()
```

7. Bivariate Analysis

Visualization 1: Contract Type vs Customer Churn

```python
# Plotting Contract Type vs Customer Churn
plt.figure(figsize=(10,5))
sns.countplot(x="Contract", hue="Churn", data=df, palette="Set2")
plt.title("Contract Type vs Customer Churn")
plt.xlabel("Contract Type")
plt.ylabel("Number of Customers")
plt.legend(title="Churn", labels=["No", "Yes"])
plt.show()
```

Visualization 2: Monthly Charges vs Customer Churn

```python
# Plotting Monthly Charges vs Customer Churn

plt.figure(figsize=(8,5))
sns.boxplot(x="Churn", y="MonthlyCharges", data=df, palette="Set1")
plt.title("Monthly Charges vs Customer Churn")
plt.xlabel("Churn")
plt.ylabel("Monthly Charges")
plt.xticks([0,1],["No","Yes"])
plt.show()
```

Visualization 3: Customer Tenure vs Customer Churn

```python
# Plotting Customer Tenure vs Customer Churn
plt.figure(figsize=(8,20))
sns.boxplot(x="Churn", y="tenure", data=df, palette="Pastel1")
plt.title("Customer Tenure vs Customer Churn")
plt.xlabel("Churn")
plt.ylabel("Tenure (Months)")
plt.xticks([0,1],["No","Yes"])
plt.show()
```
